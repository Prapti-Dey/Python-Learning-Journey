# **Output Formatting using print()**

```python
name = "Prapti Dey"
city = "Dhaka"

print("Hello, my name is", name, ". I live in", city)
```

```python
print("Hello, my name is", name, ". I live in", city, sep = "-")
```

```python
print(f"Hello, my name is {name}. I live in {city}.") # f-string
```

```python
print("Hello, my name is {name}. I live in {city}.")
```

```python
day = "7"
month = "11"
year = "2026"

print(day, month, year, sep = "-")
```

```python
print("Name", "City", sep = " | ")
print("-" * 13)
print("Prapti", "Dhaka", sep = " | ")
print("Rima", "Dhaka", sep = " | ")
```

```python
print("Name\t", "\tCity", sep = " |")
print("-" * 22)
print("Prapti\t", "\tDhaka", sep = " |")
print("Rima\t", "\tDhaka", sep = " |")
```

```python
print("Hello World\nHello Bangladesh")
print("\nHello World\tHello Bangladesh")

# \n represents a newline character and is an escape sequence.
# \t represents a tab character
```

```python
print("Name \tCity")
print("-" * 14)
print("Prapti \tDhaka")
print("Rima \tDhaka")
```
